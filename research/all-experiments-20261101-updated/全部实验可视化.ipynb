# 先导杯：最近全部测试可视化
14次独立记录，新增PR22同机三版对照，重复备份不重复计数。诊断与正常测试分开标识，历史不同实例不据此认定加速。

从上到下运行代码格。图可直接查看；数据读取 → 数组计算 → 画图，与课程练习对应。

```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = ['Microsoft YaHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
# data.json 与本笔记本放在同一目录
runs = json.loads(Path('data.json').read_text(encoding='utf-8'))['runs']
times = np.array([[v['time_s'] for v in r['result']['loads']] for r in runs])
rates = np.array([[v['consistency_rate'] * 100 for v in r['result']['loads']] for r in runs])
labels = ['10/28 FP32', '10/28 FP16', '10/29 FP32 首轮', '10/29 FP32 补跑', '10/29 cProfile诊断', '10/29 RoPE', '10/29 同晚 FP32', '10/29 编译 第1轮', '10/29 编译 第2轮', '11/01 PR20 新版', '11/01 PR20前版对照', '11/01 ④KV预分配', '11/01 PR22前版对照', '11/01 ⑤去冗余操作']
loads = ['负载1：B1 L32 G256', '负载2：B8 L32 G256', '负载3：B4 L32 G512', '负载4：B4 L512 G64']
```

## 1. 全部耗时
每根柱是一轮完整记录的一个负载。横轴从0开始；灰色为cProfile诊断，含诊断开销，不能作正式性能成绩。跨日期仅用于回顾。首轮FP32缺配对日志，其余13轮均与各自日志匹配。

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 13), layout='constrained')
colors = ['#52657c'] * len(runs)
colors[1] = '#bd5952'
colors[4] = '#aaaaaa'
for i in [5, 7, 8]: colors[i] = '#248980'
colors[9] = '#de982b'
colors[11] = '#389c96'
colors[13] = '#126c68'
for k, ax in enumerate(axes.flat):
    ax.barh(labels, times[:, k], color=colors)
    ax.invert_yaxis()
    ax.set_xlim(0, times[:, k].max()*1.18)
    ax.set_title(loads[k])
    ax.set_xlabel('耗时 / 秒（越小越好）')
    for j, v in enumerate(times[:, k]): ax.text(v, j, f' {v:.3f}', va='center', fontsize=9)
fig.suptitle('全部14轮测试：历史耗时回顾', fontsize=15)
plt.show()
```

## 2. 输出一致率
颜色与数字表示和公开FP32参考的匹配率，不是模型预测准确率。直接FP16在负载2、3约69%；其他记录四项均100%。

```python
fig, ax = plt.subplots(figsize=(10, 8.5), layout='constrained')
im = ax.imshow(rates, vmin=60, vmax=100, cmap='YlGnBu', aspect='auto')
ax.set_yticks(np.arange(len(labels)), labels)
ax.set_xticks(np.arange(4), ['负载1', '负载2', '负载3', '负载4'])
for j in range(len(labels)):
    for k in range(4): ax.text(k, j, f'{rates[j,k]:.2f}%', ha='center', va='center', color='white' if rates[j,k]>90 else 'black')
fig.colorbar(im, ax=ax, label='一致率 / %')
ax.set_title('全部14轮：与公开参考输出的一致率')
plt.show()
```

## 3. 10月29日同晚版本对照
用同晚原版作参照：加速比＝原版耗时÷候选耗时，大于1表示更快。两轮编译结果独立显示，不挑单项最好值；单轮/两轮证据不等于稳定性定论。

```python
baseline = times[6]
selected = [5, 7, 8]
versions = ['仅RoPE', 'RoPE+编译 第1轮', 'RoPE+编译 第2轮']
x = np.arange(4)
fig, ax = plt.subplots(figsize=(10, 5), layout='constrained')
for j, (idx, name) in enumerate(zip(selected, versions)):
    values = baseline / times[idx]
    bars = ax.bar(x+(j-1)*0.24, values, width=0.24, label=name)
    ax.bar_label(bars, fmt='%.2f×', padding=3)
ax.axhline(1, color='gray', linestyle='--')
ax.set_xticks(x, ['负载1', '负载2', '负载3', '负载4'])
ax.set_ylim(0, 4.5)
ax.set_ylabel('相对同晚FP32的加速比')
ax.set_title('已有优化的观测收益（不包含编译等待总时长）')
ax.legend()
plt.show()
```

## 4. 11月1日：PR20是否继续提速
同一实例、各一轮。变化＝（新版耗时÷前版耗时−1）×100%；正数表示新版更慢。这张图专门放大小差异，不能替代第一张图的绝对耗时视图。

```python
change = (times[9] / times[10] - 1) * 100
fig, ax = plt.subplots(figsize=(9, 4.5), layout='constrained')
bars = ax.bar(['负载1', '负载2', '负载3', '负载4'], change, color='#de982b')
ax.bar_label(bars, labels=[f'+{v:.3f}%' for v in change], padding=4)
ax.axhline(0, color='gray')
ax.set_ylim(-0.4, 0.4)
ax.set_ylabel('新版耗时变化 / %（正数为变慢）')
ax.set_title('PR20本轮未观察到新增收益；差距很小，不能认定稳定变慢')
plt.show()
```

## 5. PR22：新一轮同机对照
版本④在位置表＋编译基础上加入KV预分配；⑤再去除冗余类型转换和跳过p=0的Dropout。左图看绝对耗时，右图两者均与同机前版比较。各一轮、四项均100%一致；无误差条，不代表稳定性已验证。⑤相对④的额外耗时降低为2.024%、2.433%、2.067%、1.558%。

```python
x = np.arange(4)
fig, axes = plt.subplots(1, 2, figsize=(13, 5), layout='constrained')
for j, (idx, name, color) in enumerate(zip([12, 11, 13], ['前版：位置表＋编译', '④：加入KV预分配', '⑤：再去冗余操作'], ['#52657c', '#389c96', '#126c68'])):
    bars = axes[0].bar(x+(j-1)*0.25, times[idx], width=0.25, label=name, color=color)
    axes[0].bar_label(bars, fmt='%.3f', fontsize=8, padding=3)
axes[0].set_ylim(0, 7.3)
axes[0].set_ylabel('耗时 / 秒（越小越好）')
axes[0].set_title('PR22 同机三版绝对耗时')
axes[0].legend(fontsize=8, loc='upper left')
for j, (idx, name, color) in enumerate(zip([11, 13], ['④相对同机前版', '⑤相对同机前版'], ['#389c96', '#126c68'])):
    reduction = (1-times[idx]/times[12])*100
    bars = axes[1].bar(x+(j-0.5)*0.32, reduction, width=0.32, label=name, color=color)
    axes[1].bar_label(bars, fmt='%.2f%%', fontsize=9, padding=3)
axes[1].set_ylim(0, 17)
axes[1].set_ylabel('耗时降低 / %（越大越好）')
axes[1].set_title('相对同机前版的观测收益（各一轮）')
axes[1].legend(fontsize=9)
for ax in axes: ax.set_xticks(x, ['负载1', '负载2', '负载3', '负载4'])
plt.show()
```

## 结论与下一步

- RoPE＋编译：已有明显的计时收益，但首次编译等待仍需单独处理。
- PR20：两版四项100%一致；新版本本轮慢约0.24%～0.27%，暂不替换前版。
- PR22：④比同机前版耗时降低10.36%～11.65%；⑤比④再降低1.56%～2.43%，⑤比前版降低11.93%～13.81%。各一轮，稳定性待验证。
